# Titanic Dataset - Exploratory Data Analysis (EDA)


## Introduction  
This notebook performs **Exploratory Data Analysis (EDA)** on the Titanic dataset.  
We aim to explore data distributions, relationships, and patterns to understand survival trends.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")

train.head()
```

## Data Overview

```
# Info and basic statistics
train.info()
train.describe(include="all")
```

## Missing Values

```
train.isnull().sum()
```

## Value Counts for Categorical Variables

```
train['Sex'].value_counts(), train['Embarked'].value_counts(), train['Pclass'].value_counts()
```

## Distribution of Numerical Features

```
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(train['Age'].dropna(), bins=30, kde=True, ax=axes[0])
axes[0].set_title("Age Distribution")

sns.histplot(train['Fare'], bins=30, kde=True, ax=axes[1])
axes[1].set_title("Fare Distribution")
plt.show()
```

## Survival by Categorical Features

```
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.countplot(x="Sex", hue="Survived", data=train, ax=axes[0])
axes[0].set_title("Survival by Sex")

sns.countplot(x="Pclass", hue="Survived", data=train, ax=axes[1])
axes[1].set_title("Survival by Pclass")

sns.countplot(x="Embarked", hue="Survived", data=train, ax=axes[2])
axes[2].set_title("Survival by Embarked")
plt.show()
```

## Boxplots of Age and Fare vs Survival

```
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(x="Survived", y="Age", data=train, ax=axes[0])
axes[0].set_title("Age vs Survival")

sns.boxplot(x="Survived", y="Fare", data=train, ax=axes[1])
axes[1].set_title("Fare vs Survival")
plt.show()
```

## Correlation Heatmap

```
corr = train.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

## Pairplot of Selected Features

```
sns.pairplot(train[['Survived','Age','Fare','Pclass']], hue="Survived")
plt.show()
```


## Key Insights & Observations
1. **Sex**: Females had a much higher survival rate than males.  
2. **Pclass**: Higher-class passengers (1st class) had better chances of survival.  
3. **Age**: Younger children had higher survival rates compared to older passengers.  
4. **Fare**: Passengers with higher fares were more likely to survive, likely linked to class.  
5. **Embarked**: Passengers from port 'C' had slightly higher survival compared to 'S'.  

## Final Summary
The EDA suggests that **gender, passenger class, and fare** played crucial roles in survival.  
Women, children, and wealthier passengers had a better chance of survival.
